```python

from pathlib import Path
import re
import pandas as pd

# ==================================================
# 1. PROJECT FOLDERS
# ==================================================

BASE_DIR = Path(
    r"C:\Users\Tanvi\OneDrive\Desktop\Megaproject"
)

DATA_DIR = BASE_DIR / "datasets"
OUTPUT_DIR = BASE_DIR / "cleaned_data"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Project folder:", BASE_DIR)
print("Dataset folder:", DATA_DIR)
print("Output folder:", OUTPUT_DIR)

if not DATA_DIR.exists():
    raise FileNotFoundError(
        f"Dataset folder not found: {DATA_DIR}"
    )


# ==================================================
# 2. HELPER FUNCTIONS
# ==================================================

def clean_column_name(name):
    """Standardize a column name."""
    name = str(name).strip().lower()
    name = name.replace("%", "percent")

    name = re.sub(r"[^a-z0-9]+", "_", name)
    name = name.strip("_")

    return name if name else "unnamed_column"


def make_unique_columns(columns):
    """Ensure standardized column names are unique."""
    seen = {}
    result = []

    for col in columns:
        count = seen.get(col, 0) + 1
        seen[col] = count

        if count == 1:
            result.append(col)
        else:
            result.append(f"{col}_{count}")

    return result


def safe_filename(name):
    """Make a filename safe for Windows."""
    name = re.sub(r'[<>:"/\\|?*]', "_", name)
    name = re.sub(r"\s+", "_", name)
    return name.strip("._")


def read_source_file(file_path):
    """
    Detect HTML, modern Excel, or legacy Excel.
    Returns a list of (table_name, DataFrame) pairs.
    """

    with open(file_path, "rb") as f:
        signature = f.read(4096).lstrip().lower()

    # HTML reports sometimes use the .xls extension.
    is_html = (
        signature.startswith(b"<!doctype html")
        or signature.startswith(b"<html")
        or b"<table" in signature
    )

    # Modern .xlsx/.xlsm files are ZIP-based and
    # commonly begin with the PK signature.
    is_modern_excel = signature.startswith(b"pk")

    if is_html:
        tables = pd.read_html(file_path)

        if not tables:
            raise ValueError("No HTML tables found.")

        return [
            (f"table_{i + 1}", table)
            for i, table in enumerate(tables)
        ]

    elif is_modern_excel or file_path.suffix.lower() in {
        ".xlsx", ".xlsm"
    }:
        sheets = pd.read_excel(
            file_path,
            sheet_name=None,
            engine="openpyxl"
        )

        return list(sheets.items())

    elif file_path.suffix.lower() == ".xls":
        sheets = pd.read_excel(
            file_path,
            sheet_name=None,
            engine="xlrd"
        )

        return list(sheets.items())

    else:
        raise ValueError(
            f"Unsupported file type: {file_path.suffix}"
        )


# ==================================================
# 3. CLEAN ONE DATAFRAME
# ==================================================

def clean_dataframe(df):
    """Perform safe initial cleaning."""

    original_shape = df.shape

    # Remove completely empty rows and columns.
    df = df.dropna(axis=0, how="all").copy()
    df = df.dropna(axis=1, how="all").copy()

    # Standardize column names.
    df.columns = make_unique_columns(
        [clean_column_name(c) for c in df.columns]
    )

    # Count missing values.
    missing_values = (
        df.isnull()
        .sum()
        .to_dict()
    )

    missing_values = {
        col: int(count)
        for col, count in missing_values.items()
        if count > 0
    }

    # Detect exact duplicate rows, but do not delete them.
    duplicate_count = int(df.duplicated().sum())

    # Record the number of fully empty rows/columns removed.
    report = {
        "original_rows": original_shape[0],
        "original_columns": original_shape[1],
        "cleaned_rows": df.shape[0],
        "cleaned_columns": df.shape[1],
        "duplicate_rows_found": duplicate_count,
        "missing_values_by_column": str(missing_values),
        "columns_after_cleaning": str(df.columns.tolist())
    }

    return df, report


# ==================================================
# 4. PROCESS EACH FILE SEPARATELY
# ==================================================

supported_extensions = {
    ".xls", ".xlsx", ".xlsm", ".html", ".htm"
}

source_files = sorted(
    file for file in DATA_DIR.iterdir()
    if file.is_file()
    and file.suffix.lower() in supported_extensions
    and not file.name.startswith("~$")
)

if not source_files:
    raise FileNotFoundError(
        f"No supported files found in {DATA_DIR}"
    )

all_reports = []

for file_path in source_files:
    print("\n" + "=" * 60)
    print("Processing:", file_path.name)

    try:
        tables = read_source_file(file_path)

        for table_name, original_df in tables:
            print("\nTable/sheet:", table_name)

            cleaned_df, report = clean_dataframe(
                original_df.copy()
            )

            # Give every source file and sheet its own CSV.
            output_name = safe_filename(
                f"{file_path.stem}_{table_name}_cleaned"
            ) + ".csv"

            output_path = OUTPUT_DIR / output_name

            cleaned_df.to_csv(
                output_path,
                index=False,
                encoding="utf-8-sig"
            )

            report["source_file"] = file_path.name
            report["table_or_sheet"] = table_name
            report["output_file"] = output_name
            report["status"] = "SUCCESS"

            all_reports.append(report)

            print("Original shape:", (
                report["original_rows"],
                report["original_columns"]
            ))
            print("Cleaned shape:", cleaned_df.shape)
            print("Duplicate rows found:",
                  report["duplicate_rows_found"])
            print("Saved to:", output_path)

    except Exception as error:
        print("ERROR:", error)

        all_reports.append({
            "source_file": file_path.name,
            "table_or_sheet": "",
            "status": "FAILED",
            "error": str(error)
        })


# ==================================================
# 5. SAVE A CLEANING SUMMARY REPORT
# ==================================================

summary_df = pd.DataFrame(all_reports)

summary_path = OUTPUT_DIR / "cleaning_summary.csv"

summary_df.to_csv(
    summary_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 60)
print("CLEANING PROCESS FINISHED")
print("Files processed:", len(source_files))
print("Output folder:", OUTPUT_DIR)
print("Summary report:", summary_path)

print("\nProcessing status:")
if "status" in summary_df.columns:
    print(summary_df["status"].value_counts())
```
